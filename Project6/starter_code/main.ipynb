```python
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm

from lidar.surfaces import surfG, gradG, A_
from lidar.simulation import lidarsim, plot_surfaces

print("Modules imported successfully!")
print(f"Surface amplitudes: {A_}")
```

## Verify Surface Functions

Test that the surface and gradient functions are working correctly.

```python
test_x = 0.0
test_y = 0.0
test_A = A_[0]

z_test = surfG(test_x, test_y, test_A)
print(f"Surface height at ({test_x}, {test_y}): z = {z_test:.4f}")

grad_test = gradG(test_x, test_y, test_A)
print(f"Surface gradient at ({test_x}, {test_y}): [{grad_test[0]:.4f}, {grad_test[1]:.4f}]")
```

## Visualize Test Surfaces

Create 3D plots of the surfaces with different amplitudes.

```python
# Create mesh grid
x = y = np.arange(-0.5, 0.5 + 0.01, 0.01)
X, Y = np.meshgrid(x, y)

for i in range(len(A_)):
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    # Compute surface heights
    zs = np.array(surfG(np.ravel(X), np.ravel(Y), A_[i]))
    Z = zs.reshape(X.shape)
    
    # Plot surface
    surf = ax.plot_surface(X, Y, Z, cmap=cm.plasma, alpha=0.8)
    
    ax.set_xlim(-0.75, 0.75)
    ax.set_ylim(-0.75, 0.75)
    ax.set_zlim(1.5, 4.0)
    ax.set_xlabel('X (m)', fontsize=12)
    ax.set_ylabel('Y (m)', fontsize=12)
    ax.set_zlabel('Z (m)', fontsize=12)
    ax.set_title(f'Surface Geometry - Amplitude A={A_[i]}', fontsize=14)
    ax.view_init(elev=30., azim=45)
    
    fig.colorbar(surf, ax=ax, label='Height (m)')
    plt.tight_layout()
    plt.show()

print("Surface visualization complete!")
```

## Run Simulations

Execute the LIDAR ray-tracing simulation for each surface, at two different lens focal lengths with the same pixel pitch and number of pixels. A longer focal length gives a narrower field of view but smaller pixel footprints.

For each surface and focal length the figure shows the 3D point cloud (with the lens aperture as a red circle and the sensor as a blue square, to scale) next to the reconstructed height map. 

```python
# Provided: run the simulation for each surface at two focal lengths and compare them.
# This calls lidarsim() for every (surface, focal length) pair and plots the results.
from lidar.reference import focal_length_study

focal_lengths = [0.1, 0.2]       # lens focal lengths (m), same pixel pitch

print("=" * 70)
print("Starting LIDAR Simulations")
print("=" * 70)

for surface_idx in range(len(A_)):
    print(f"\nRunning simulation for surface {surface_idx + 1} of {len(A_)}...")
    print(f"Surface amplitude A = {A_[surface_idx]}")

    # Run the simulation for each focal length and plot the comparison
    focal_length_study(surface_idx, focal_lengths)

print("\n" + "=" * 70)
print("All simulations completed!")
```
